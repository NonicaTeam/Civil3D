```python
def parse_tree(node):
    """Recursively parse HTML tree into a FancyTree structure."""

    title_tag = node.find("a")
    if not title_tag:
        return None
    
    subnodes = node.find(attrs={"class": "treeSubnodesHidden"})

    tree_node = {
        "title": title_tag.get("title"),
        "href": title_tag.get("href") if title_tag.get("href") else None,
        "folder": subnodes is not None
    }
    if subnodes is not None:
        extract_nodes = subnodes.find(attrs={"class": "treeNode"}) 
        tree_node["children"] = parse_tree(extract_nodes)
    return tree_node
```

```python
def parse_tree(node):
    """Recursively parse HTML tree into a FancyTree structure."""

    title_tag = node.find("a")
    if not title_tag:
        return None
    
    subnodes = node.find(attrs={"class": "treeSubnodesHidden"})

    tree_node = {
        "title": title_tag.get("title"),
        "href": title_tag.get("href") if title_tag.get("href") else None,
        "folder": subnodes is not None
    }
    if subnodes is not None:
        children = subnodes.find_all(attrs={"class": "treeNode"}, recursive=False) 
        list_children = []
        for child in children:
            list_children.append(parse_tree(child))
        tree_node["children"] = list_children
    return tree_node
```

```python
root_node = body.find(attrs={"class": "treeNode"})
parsed_tree = parse_tree(root_node)
json_output = json.dumps(parsed_tree, indent=4)
print(json_output)
```
