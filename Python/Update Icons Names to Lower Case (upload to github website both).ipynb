```python
import os
import shutil

# Replace with your source folder's absolute path
source_folder = r"C:\Users\jamo1\Desktop\iconsfolder"

# Replace with your destination folder's absolute path
destination_folder = r"C:\Users\jamo1\Desktop\iconsfolder/updated"

# Create destination folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)

# Iterate through each item in the source folder
for filename in os.listdir(source_folder):
    source_file = os.path.join(source_folder, filename)
    
    # Check if the item is a file (ignores directories)
    if os.path.isfile(source_file):
        # Convert the filename to lowercase
        new_filename = filename.lower()
        destination_file = os.path.join(destination_folder, new_filename)
        
        # Copy the file to the destination folder with the new name
        shutil.copy2(source_file, destination_file)

print("All files have been copied with lowercase names to:", destination_folder)
```
